### Comparison of Harris Part I and Part III

Both datasets contain 157 globular cluster. The ID columns are identical and in the same order, while the two datasets contain different physical parameters.

```python
harris = pd.merge(harris1, harris3, on="ID")
```

```python
harris.head()
```

```python
harris[["ID", "L", "B", "R_gc", "X", "Y", "Z", "v_r", "v_LSR"]].head(10)
```

```python
harris[["L", "B", "R_gc", "X", "Y", "Z", "v_r", "v_LSR"]].isnull().sum()
```

```python
harris[["v_r","v_LSR"]].describe()
```

```python
import matplotlib.pyplot as plt

plt.scatter(harris["L"], harris["v_LSR"])
plt.xlabel("Galactic Longitude L (degrees)")
plt.ylabel("v_LSR (km/s)")
plt.title("Global Cluster Velocity vs Galactic Longitude")
plt.show()
```

```python
harris["L_centered"] = ((harris["L"] + 180) % 360) - 180
```

```python
plt.scatter(harris["L_centered"], harris["v_LSR"])
plt.xlabel("Galactic Longitude L (degrees)")
plt.ylabel("v_LSR (km/s)")
plt.title("Global Cluster Velocity vs Galactic Longitude (Centered)")
plt.show()
```

```python
left = harris[harris["L_centered"] < 0]
right = harris[harris["L_centered"] > 0]
```

```python
print("L < 0 mean v_LSR:", left["v_LSR"].mean())
print("L > 0 mean v_LSR:", right["v_LSR"].mean())
```

```python
opposite = harris[
((harris["L_centered"] < 0) & (harris["v_LSR"] < 0))|
((harris["L_centered"] > 0) & (harris["v_LSR"] > 0))
]

opposite[["ID", "L_centered", "v_LSR"]]
```

The clusters show an overall kinematic pattern, but using only the signs of longitude and velocity identifirs too many unusual clusters.

```python
import numpy as np

harris["sin_L"] = np.sin(np.radians(harris["L_centered"]))
```

```python
plt.scatter(harris["sin_L"], harris["v_LSR"])
plt.xlabel("sin(L)")
plt.ylabel("v_LSR (km/s)")
plt.title("Global Cluster Velocity vs sin(L)")
plt.show()
```

```python
valid = harris[["sin_L", "v_LSR"]].dropna()

fit = np.polyfit(valid["sin_L"], valid["v_LSR"], 1)
fit
```

```python
x = np.linspace(-1, 1, 100)
y = fit[0] * x + fit[1]

plt.scatter(harris["sin_L"], harris["v_LSR"])
plt.plot(x, y)

plt.xlabel("sin(L)")
plt.ylabel("v_LSR (km/s)")
plt.title("Global Cluster Velocity and Bulk Trend")
plt.show()
```

```python
harris["v_predicted"] = fit[0] * harris["sin_L"] + fit[1]
harris["v_residual"] = harris["v_LSR"] - harris["v_predicted"]

harris[["ID", "L_centered", "v_LSR", "v_predicted", "v_residual"]].sort_values(
    "v_residual", key=abs, ascending=False
).head(10)
```

```python
outliers = harris.loc[harris["v_residual"].abs().nlargest(5).index]

plt.scatter(harris["sin_L"], harris["v_LSR"])
plt.plot(x, y)

for _, row in outliers.iterrows():
    plt.text(row["sin_L"], row["v_LSR"], row["ID"])

plt.xlabel("sin(L)")
plt.ylabel("v_LSR (km/s)")
plt.title("Global Cluster Velocity and kinematic outliers")
plt.show()
```

Five clusters show the largest deviations from the bulk velocity trend: NGC 3201, NGC 6101, NGC 6287, Djong 1 and NGC 6934. These clusters are considered kinematic outliers for further investigation.
