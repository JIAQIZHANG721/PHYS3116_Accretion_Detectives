```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load spatial data from Harris catalogue
harris1 = pd.read_csv("HarrisPartI.csv")

# Check the positions of globular clusters
harris1[["ID", "X", "Y", "Z", "R_gc"]].head()
```

```python
# Plot the spatial distribution of globular clusters

plt.figure(figsize=(8, 7))

plt.scatter(harris1["X"], harris1["Y"], s=25, alpha=0.6, label="Globular Clusters")

# Mark the Galactic Center
plt.scatter(0, 0, color='red', marker='*', s=180, label="Galactic Center")

plt.xlabel("X (kpc)")
plt.ylabel("Y (kpc)")
plt.title("Spatial Distribution of Milky Way Globular Clusters")

plt.axhline(0, color='gray', linestyle='--', linewidth=0.4)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.4)

plt.axis('equal')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Select the top five candidates from Week 4
candidate_names = ["NGC 1851", "Terzan 7", "NGC 3201", "NGC 6934", "Pal 12"]

candidates = harris1[harris1["ID"].isin(candidate_names)]

# Check candidate positions
candidates[["ID", "X", "Y", "Z", "R_gc"]]
```

```python
# Plot all globular clusters
plt.figure(figsize=(9,7))

plt.scatter(harris1["X"], harris1[ "Y"], color="lightgray", s=25, alpha=0.6, label="All Globular Clusters")

# Highlight selected candidates
plt.scatter(candidates["X"], candidates["Y"], color="red", s=80, marker="o", label="Selected Candidates")

# Add candidate names
for _, cluster in candidates.iterrows():
    plt.annotate(cluster["ID"], (cluster["X"], cluster["Y"]), xytext=(5,5), textcoords="offset points", fontsize=9)

# Mark the Galactic centre
plt.scatter(0, 0, color="black", marker="*", s=180, label="Galactic Centre")

plt.xlabel("X (kpc)")
plt.ylabel("Y (kpc)")
plt.title("Spatial Distribution of Selected Globular Cluster Candidate")

plt.axhline(0, color="gray", linestyle="--", alpha=0.4)
plt.axvline(0, color="gray", linestyle="--", alpha=0.4)

plt.axis("equal")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Plot the X-Z spatial distribution

plt.scatter(harris1["X"], harris1[ "Z"], color="lightgray", s=25, alpha=0.6, label="All Globular Clusters")

plt.scatter(candidates["X"], candidates["Z"], color="red", s=80, label="Selected Candidates")

for _, cluster in candidates.iterrows():
    plt.annotate(cluster["ID"], (cluster["X"], cluster["Z"]), xytext=(5,5), textcoords="offset points", fontsize=9)

plt.scatter(0, 0, color="black", marker="*", s=180, label="Galactic Centre")

plt.xlabel("X (kpc)")
plt.ylabel("Z (kpc)")
plt.title("X-Z Distribution of Globular Cluster Candidates")

plt.axis("equal")
plt.legend()
plt.grid(alpha=0.3)
plt.xlim(-30, 40)
plt.ylim(-25, 20)
plt.show()
```

## Spatial Analysis

The five candidates are distributed across different regions of the Milky Way. Pal 12 is the furthest from the Galactic plane among them.
